# 1 Bases de python


Python a été conçu par Guido Van Rossum. Ce qui a commencé comme un projet amateur est rapidement devenu un langage de programmation polyvalent : aujourd'hui, Python peut être utilisé pour développer pratiquement n'importe quel type de logiciel. 

Tout d'abord, Python est open source. Son utilisation est gratuite. Deuxièmement, il est très facile de créer des paquets en Python, qui sont des codes que vous pouvez partager avec d'autres personnes pour résoudre des problèmes spécifiques. Au fil du temps, de plus en plus de ces paquets spécialement conçus pour la science des données ont été développés. Les gens qualifient souvent Python de « couteau suisse » des langages de programmation, car il permet de réaliser presque toutes les tâches. 

Le shell Python est un endroit où vous pouvez saisir du code Python et voir immédiatement les résultats.

Lorsqu'on saisit par exemple 1+1 et qu'on appuie sur entrée, Python interprète ce qu'on a saisi et affiche le résultat du calcul. 


```python
1 + 1
```

Outre le travail interactif avec Python, on peut également exécuter Python dans des scripts Python. Ce sont de simples fichiers texte avec l'extension *.py*. Il s'agit essentiellement d'une liste de commandes Python qui sont exécutées, comme si nous tapions nous-même les commandes dans le shell, ligne par ligne.
En plaçant le code dans des scripts Python plutôt que de retaper manuellement chaque étape de manière interactive, on conserve une structure claire, ce qui évite de tout retaper à chaque fois qu'on souhaite apporter une modification. 


```python
# Addition, subtraction
print(5 + 5)
print(5 - 5)

# Multiplication, division, modulo, and exponentiation
print(3 * 5)
print(10 / 2)
print(18 % 7)
print(4 ** 2)
```
